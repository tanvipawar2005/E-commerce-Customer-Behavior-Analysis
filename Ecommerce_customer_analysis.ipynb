```python
# Outlier detection using boxplot in total spending
sns.boxplot(x=data['Total Spent (Rs)'])
plt.title("Outliers in Total Spending")
plt.xlabel("Total Spending (Rs)")
plt.ylabel("Distribution")
plt.show()
```

# **Bivariate/ Multivariate Graphical analysis**

```python
# Monthly Income vs Total Spending by Gender and Customer Rating
plt.title("Monthly Income vs Total Spending by Gender and Customer Rating")
sns.scatterplot(x='Monthly Income (Rs)', y='Total Spent (Rs)',
                size='Customer Rating', hue='Gender', data=data, alpha=0.7)
plt.show()
```

```python
# relationship between Monthly Income and Total spending with regression line
sns.jointplot(x='Monthly Income (Rs)', y='Total Spent (Rs)',height=5, data=data, kind='reg')
```

# **Basic Descriptive Questions**

```python
# 1. How many male and female customers are in the dataset?
gender_count = data['Gender'].value_counts()
sns.barplot(x=gender_count.index, y=gender_count.values, palette="viridis")
plt.title("Male vs Female Customers")
plt.ylabel("Count")
plt.show()
print("Gender Count:", gender_count)
```

```python
# 2. What is the average age of all customers?
avg_age = data['Age'].mean()
plt.figure()
sns.histplot(data['Age'], bins=10, kde=True, color='skyblue')
plt.axvline(avg_age, color='red', linestyle='--', label=f'Avg Age = {avg_age:.2f}')
plt.title("Age Distribution")
plt.legend()
plt.show()
print(f"The average customer age is {avg_age:.2f} years, with most customers between 25-50.")
```

```python
# 3.Which product category has the highest number of purchases?
purchase_by_category = data.groupby('Product Category')['Purchases'].sum()
purchase_by_category.plot(kind='bar', color='teal')
plt.title("Total Purchases by Product Category")
plt.ylabel("Purchases")
plt.show()
print(f" {purchase_by_category.idxmax()} has the highest total purchases which is {purchase_by_category.max()}.")
```

```python
# 4. Which platform has the most users?
platform_users = data['Platform Used'].value_counts()
explode = [0.1 if i == platform_users.idxmax() else 0 for i in platform_users.index]
plt.figure()
plt.pie(platform_users, labels=platform_users.index, autopct='%1.1f%%', startangle=0, explode=explode)
plt.title("Platform Popularity")
plt.show()
print(f" {platform_users.idxmax()} is the most used platform.")
```

```python
# 5. Which customer has the highest total spending?
top_5 = data.nlargest(5, 'Total Spent (Rs)')
sns.barplot(x='Total Spent (Rs)', y='Customer ID', data=top_5, palette='Reds')
plt.title("Top 5 Highest Spending Customers")
plt.xlabel("Total Spent (Rs)")
plt.ylabel("Customer ID")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()

print("The highest spending customer is:", top_5.iloc[0]['Customer ID'],
      "with ₹", top_5.iloc[0]['Total Spent (Rs)'])
```

```python
# 6. What is the average monthly income of customers who use the Meesho App?
meesho_users = data[data['Platform Used'] == 'Meesho']
avg_income_meesho = meesho_users['Monthly Income (Rs)'].mean()

print("The average monthly income of Meesho App users is: ₹", round(avg_income_meesho, 2))
```

```python
# 7. Which customer has the longest average session duration?
longest_session = data.loc[data['Session Duration (mins)'].idxmax()]

print("Customer", longest_session['Customer ID'],
      "has the longest session of", longest_session['Session Duration (mins)'], "mins.")
```

```python
#8 How many customers have a customer rating of 5?
counts = data['Customer Rating'].value_counts().sort_index()
plt.figure()
plt.pie(counts, labels=[f"{i} star" for i in counts.index], autopct='%1.1f%%')
plt.title("Customer Rating Distribution")
plt.show()

print(f"{(data['Customer Rating']==5).sum()} customers gave a perfect rating of 5.")
```

```python
# 9. What is the total amount spent by all customers?
total_spent_all = data['Total Spent (Rs)'].sum()
total_spent_all
print(f"Total amount spent by all customers is {total_spent_all} Rs.")
```

```python
# 10. What is the average spending score of all customers?
avg_spending_score = data['Spending Score'].mean()
avg_spending_score
print(f" Average spending score is {avg_spending_score:.2f}.")
```

# **Analytical / Comparative Questions**

```python
# 1. Compare average total spending between male and female customers.
avg_spent_gender = data.groupby('Gender')['Total Spent (Rs)'].mean()
plt.figure()
avg_spent_gender.plot(kind='bar', color=['#3498db','#e74c3c'])
plt.title("Average Total Spending by Gender")
plt.ylabel("Average Total Spent (Rs)")
plt.grid(axis='y', alpha=0.6)
plt.show()
print(f"Males spend Rs.{avg_spent_gender['Male']:.2f}, Females spend Rs.{avg_spent_gender['Female']:.2f} on average.")
```

```python
# 2. Identify the most popular product category based on number of purchases.
category_purchases = data.groupby("Product Category")["Purchases"].sum()
category_purchases.plot(kind="bar")
plt.title("Most Popular Product Category (by Purchases)")
plt.xlabel("Product Category")
plt.ylabel("Total Purchases")
plt.xticks(rotation=45)
plt.show()

print(f" The most popular product category is '{category_purchases.idxmax()}' with {category_purchases.max()} total purchases.")
```

```python
# 3	Which age group has the highest spending score?
data['Age Group (10yrs)'] = pd.cut(data['Age'], bins=range(0, 101, 10))
avg_score_age2 = data.groupby('Age Group (10yrs)')['Spending Score'].mean()

avg_score_age2.plot(kind='line', marker='o', color='green')
plt.title("Average Spending Score by 10-Year Age Groups")
plt.ylabel("Spending Score")
plt.xlabel("Age Group")
plt.grid(alpha=0.5)
plt.show()

print(f"Age group '{avg_score_age2.idxmax()}' has the highest spending score {avg_score_age2.max():.2f}")
```

```python
# 4 What is the relationship between average purchase value and total spent?
plt.scatter(data['Avg Purchase Value (Rs)'], data['Total Spent (Rs)'])
plt.title("Relationship: Avg Purchase Value vs Total Spent")
plt.xlabel("Average Purchase Value (Rs)")
plt.ylabel("Total Spent (Rs)")
plt.show()
corr = data['Avg Purchase Value (Rs)'].corr(data['Total Spent (Rs)'])
print(f"The correlation between Avg Purchase Value and Total Spent is {corr:.2f}")
```

```python
# 5.How many customers made more than 3 purchases?
purchase_counts = ["≤ 3 Purchases", "> 3 Purchases"]
values = [(data['Purchases'] <= 3).sum(), (data['Purchases'] > 3).sum()]
plt.bar(purchase_counts, values, color=['skyblue', 'orange'])
plt.title("Customers by Purchase Count")
plt.ylabel("Number of Customers")
plt.show()

print("Number of customers who made more than 3 purchases:", values[1])
```

```python
# 6. Which platform has the highest average customer rating?
platform_ratings = data.groupby("Platform Used")["Customer Rating"].mean()
platform_ratings.plot(kind='bar', color='#66b3ff')
plt.title("Average Customer Rating by Platform")
plt.xlabel("Platform")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.ylim(0,5)  # since ratings are between 1 and 5
plt.show()

print(f" {platform_ratings.idxmax()} has the highest average customer rating of {platform_ratings.max():.2f}.")
```

```python
#7. Which product category has the highest average session duration?
category_sessions = data.groupby("Product Category")["Session Duration (mins)"].mean()

category_sessions.plot(kind='bar', color='#99ccff')
plt.title("Average Session Duration by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Session Duration (mins)")
plt.xticks(rotation=45)
plt.show()

print(f"{category_sessions.idxmax()} has the highest average session duration of {category_sessions.max()} minutes.")
```

```python
# 8. Correlation between website visits and purchases
sns.heatmap(data[['Website Visits','Purchases']].corr(), annot=True, cmap="coolwarm", center=0)
plt.title("Correlation between Website Visits and Purchases")
plt.show()
```

```python
# 9. Customers who spent > ₹4000 and rating < 4
result = data[(data['Total Spent (Rs)'] > 4000) & (data['Customer Rating'] < 4)]

plt.bar(result['Customer ID'].astype(str), result['Total Spent (Rs)'], color='orange')
plt.title("High-Spending Customers (₹>4000) with Rating < 4")
plt.xlabel("Customer ID")
plt.ylabel("Total Spent (Rs)")
plt.xticks(rotation=45)
plt.show()

print(result[['Customer ID','Total Spent (Rs)','Customer Rating']])
```

```python
# 10. Top 3 highest-spending customers and their platforms
top3 = data.nlargest(3, 'Total Spent (Rs)')[['Customer ID', 'Total Spent (Rs)', 'Platform Used']]

sns.barplot(x='Customer ID', y='Total Spent (Rs)', hue='Platform Used', data=top3, palette='Set2')
plt.title("Top 3 Highest-Spending Customers & Their Platforms")
plt.xlabel("Customer ID")
plt.ylabel("Total Spent (Rs)")
plt.legend(title="Platform Used")
plt.show()

print("Top 3 highest-spending customers and their platforms:")
print(top3)
```
